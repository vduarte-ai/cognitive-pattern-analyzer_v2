```python
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
texts = [
    "I always fail at everything",
    "I never do everything right",
    "Nothing ever works for me", 
    "This is the worst thing ever",
    "Everything is ruined",
    "This is a disaster",
    "I did okay today",
    "I enjoy spending time with friends",
    "Sometimes things fo well",
    "I am doing it wrong today",
    "I made many mistakes yesterday",
    "This is the worst situation ever",
    "My job is ruined"
]

labels = [
    "overgeneralization",
    "overgeneralization",
    "overgeneralization",
    "catastrophizing",
    "catastrophizing",
    "catastrophizing",
    "neutral",
    "neutral",
    "neutral",
    "neutral",
    "neutral",
    "catastrophizing",
    "catastrophizing"
]
```

```python
# Modulo de embeddings
model = SentenceTransformer("all-MiniLM-L6-v2")

# Convertir texto a números (vectores)
X = model.encode(texts)

# Convertir etiquetas a números
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(labels)

# Clasificador
clf = LogisticRegression(max_iter=2000)
# El modelo aprende
clf.fit(X, y)

print("Modelo entrenado")
```

```python
def predict(text):
    # se coloca el texto en el espacio
    embedding = model.encode([text])
    # el modelo decide
    prediction = clf.predict(embedding)
    label = label_encoder.inverse_transform(prediction)
    return label[0]

print(predict("I ruin everything"))
print(predict("Today was fine"))
```

```python
#Guarda clasificador
joblib.dump(clf, "classifier.pkl")

#guardar encoder (modelo de embeddings)
joblib.dump(model, "embedding_model.pkl")

joblib.dump(label_encoder, "label_encoder.pkl")

print("Modelo guardado")
```

```python
# Cargar modelo
clf_loaded = joblib.load("classifier.pkl")
model_loaded = joblib.load("embedding_model.pkl")
le_loaded = joblib.load("label_encoder.pkl")
```

```python
def predict_loaded(text):
    embedding = model_loaded.encode([text])
    prediction = clf_loaded.predict(embedding)
    label = le_loaded.inverse_transform(prediction)
    return label[0]
```
